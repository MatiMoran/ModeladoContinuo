## Ejercicio 2 (extra) - Método de la secante | Newton-Raphson modificado


El método de la secante consiste en utilizar el método de Newton-Raphson pero realizando la derivada numéricamente, es decir, el método iterativo consiste en cada paso calcular:

$$x_{n+1} = x_n-f(x_n)\frac{x_n-x_{n-1}}{f(x_{n})-f(x_{n-1})}$$

Como ahora se utiliza una derivada backward en cada paso, ya no se paga el costo de calcular la derivada en cada paso a cambio de derivar backward, lo que requiere utilizar dos puntos anteriores para calcular el siguiente.

## Ejercicio 1 - El método de Euler


Para comenzar, queremos implementar el método de Euler, que es el más sencillo en lo que respecta a la integración numérica de ecuaciones diferenciales, y consiste en iterar:

$$x_{n+1} = x_{n}+ dt \; f(t_n,x_n, \vec{\alpha})$$

donde $\dot{x} = f(t,x, \vec{\alpha})$ y $\vec{\alpha}$ contiene a todos los parámetros de los cuales podría depender la ecuación diferencial, como podrían ser una viscosidad, la frecuencia/amplitud de un forzado y muchas cosas más.

### Inciso $c$ - Estabilidad del método


En cada paso temporal de cualquier método, se comete un cierto error, que puede ir acumulandosé con el tiempo. La estabilidad de un método hace referencia a qué tan rápido se acumula ese error y, en el caso en que tienda a infinito cuando el tiempo tiende a infinito, diremos que el método es inestable, ya que la soluciòn diverge diverge y deja de ser correcta.

Consideremos la ecuación diferencial:

$$\dot{x} = \lambda x $$

Esta es la ecuación diferencial típica cuya solución conocemos y está dada por $x(t) = A e^{\lambda t}$.

La condición de estabilidad del método es:

$$\mid 1+ dt \lambda \mid  < 1 $$


Esta relación nos indica que paso temporal $dt$ debemos elegir para que el método sea estable. Es claro que para cualquier $\lambda > 0$ esta condición no se va a cumplir, independientemente del $dt$ que elijamos.
